```python
corr_matrix = df.corr()
plt.figure(figsize=(10,8))
sb.heatmap(corr_matrix, annot= True, cmap="coolwarm" ,fmt = ".2f", linewidths=0.5)
plt.title("Dataset Heatmap Correlation")
plt.show()
```

```python
#scatter plot
# for BMI and Skin Thickness
sb.scatterplot(data=df, x="BMI" , y="SkinThickness")
plt.title("Scatter plot between BMI and Skin Thickness")
plt.show()
```
