```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import t, chi2_contingency

# Load the dataset
df = pd.read_csv('/content/diabetes.csv')

# Display first 5 rows
print("First 5 rows of the dataset:")
display(df.head())

# Display shape
print("\nNumber of observations and variables:")
print(df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Display basic information
print("\nDataset information:")
df.info()

# Display descriptive statistics
print("\nDescriptive statistics:")
display(df.describe())
```

```python
# Select relevant variables
selected_data = df[['Glucose', 'BMI', 'Age', 'Outcome']]

print("Selected variables:")
display(selected_data.head())

# Divide the data into two groups based on Outcome
non_diabetic = df[df['Outcome'] == 0]
diabetic = df[df['Outcome'] == 1]

print("Number of non-diabetic individuals:", len(non_diabetic))
print("Number of diabetic individuals:", len(diabetic))

# Display group statistics
print("\nNon-diabetic group:")
display(non_diabetic[['Glucose', 'BMI', 'Age']].describe())

print("\nDiabetic group:")
display(diabetic[['Glucose', 'BMI', 'Age']].describe())
```

```python
# Calculate sample mean of Glucose
sample_mean_glucose = df['Glucose'].mean()

print("Sample mean of Glucose =", sample_mean_glucose)

# Sample mean of BMI
sample_mean_bmi = df['BMI'].mean()

print("Sample mean of BMI =", sample_mean_bmi)

# Sample mean of Age
sample_mean_age = df['Age'].mean()

print("Sample mean of Age =", sample_mean_age)
```

```python
# Calculate sample proportion of individuals with diabetes
number_diabetic = df['Outcome'].sum()
total_people = len(df)

sample_proportion = number_diabetic / total_people

print("Number of individuals with diabetes:", number_diabetic)
print("Total number of individuals:", total_people)
print("Sample proportion with diabetes:", sample_proportion)
print("Sample proportion as percentage:", sample_proportion * 100, "%")
```

```python
# 95% confidence interval for population mean of Glucose

glucose = df['Glucose'].dropna()

n = len(glucose)
sample_mean = glucose.mean()
sample_std = glucose.std(ddof=1)

# Standard error
standard_error = sample_std / np.sqrt(n)

# Critical t-value
confidence_level = 0.95
alpha = 1 - confidence_level

t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

# Margin of error
margin_of_error = t_critical * standard_error

# Confidence interval
lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print("Sample mean =", sample_mean)
print("Standard deviation =", sample_std)
print("Standard error =", standard_error)
print("t-critical value =", t_critical)
print("Margin of error =", margin_of_error)

print("\n95% Confidence Interval for population mean Glucose:")
print("Lower limit =", lower_limit)
print("Upper limit =", upper_limit)
```

```python
# 95% confidence interval for population proportion

x = df['Outcome'].sum()
n = len(df)

p_hat = x / n

# Standard error for proportion
se_proportion = np.sqrt((p_hat * (1 - p_hat)) / n)

# Z critical value for 95% confidence
z_critical = stats.norm.ppf(0.975)

# Margin of error
margin = z_critical * se_proportion

# Confidence interval
lower_proportion = p_hat - margin
upper_proportion = p_hat + margin

print("Sample proportion =", p_hat)
print("Standard error =", se_proportion)
print("Z-critical value =", z_critical)

print("\n95% Confidence Interval for population proportion:")
print("Lower limit =", lower_proportion)
print("Upper limit =", upper_proportion)

print("\nAs percentages:")
print("Lower limit =", lower_proportion * 100, "%")
print("Upper limit =", upper_proportion * 100, "%")
```

```python
# One-sample t-test

# Hypothesized population mean
hypothesized_mean = 120

# Perform one-sample t-test
t_statistic, p_value = stats.ttest_1samp(
    df['Glucose'],
    popmean=hypothesized_mean
)

print("One-Sample t-Test")
print("------------------")
print("Sample mean =", df['Glucose'].mean())
print("Hypothesized mean =", hypothesized_mean)
print("t-statistic =", t_statistic)
print("p-value =", p_value)
```

```python
# Significance level
alpha = 0.05

print("Significance level =", alpha)
print("p-value =", p_value)

if p_value < alpha:
    print("\nDecision: Reject the null hypothesis.")
    print("Conclusion: The average glucose level is significantly different from 120 mg/dL.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")
    print("Conclusion: There is insufficient evidence to conclude that the average glucose level differs from 120 mg/dL.")
```

```python
# Separate glucose values into two groups

glucose_non_diabetic = df[df['Outcome'] == 0]['Glucose']
glucose_diabetic = df[df['Outcome'] == 1]['Glucose']

# Perform independent two-sample t-test
# Welch's t-test is used because it does not assume equal variances

t_statistic, p_value = stats.ttest_ind(
    glucose_non_diabetic,
    glucose_diabetic,
    equal_var=False
)

print("Two-Sample t-Test")
print("------------------")

print("Mean glucose - Non-diabetic:",
      glucose_non_diabetic.mean())

print("Mean glucose - Diabetic:",
      glucose_diabetic.mean())

print("t-statistic =", t_statistic)
print("p-value =", p_value)
```

```python
alpha = 0.05

print("Significance level =", alpha)
print("p-value =", p_value)

if p_value < alpha:
    print("\nDecision: Reject the null hypothesis.")
    print("Conclusion: There is a statistically significant difference")
    print("in mean glucose levels between diabetic and non-diabetic individuals.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")
    print("Conclusion: There is insufficient evidence of a difference")
    print("in mean glucose levels between the two groups.")
```

```python
# Create BMI categories

df['BMI_Category'] = pd.cut(
    df['BMI'],
    bins=[0, 18.5, 25, np.inf],
    labels=['Underweight', 'Normal', 'Overweight/Obese'],
    right=False
)

# Display BMI categories
print(df[['BMI', 'BMI_Category', 'Outcome']].head(10))

# Frequency of BMI categories
print("\nBMI Category frequencies:")
print(df['BMI_Category'].value_counts())
```

```python
# Create contingency table
contingency_table = pd.crosstab(
    df['BMI_Category'],
    df['Outcome']
)

# Rename columns
contingency_table.columns = ['No Diabetes', 'Diabetes']

print("Contingency Table:")
display(contingency_table)
```

```python
# Perform Chi-square test

chi2_statistic, p_value, degrees_freedom, expected = chi2_contingency(
    contingency_table
)

print("Chi-Square Test")
print("----------------")
print("Chi-square statistic =", chi2_statistic)
print("Degrees of freedom =", degrees_freedom)
print("p-value =", p_value)

print("\nExpected frequencies:")
expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

display(expected_table)
```

```python
alpha = 0.05

print("Significance level =", alpha)
print("p-value =", p_value)

if p_value < alpha:
    print("\nDecision: Reject the null hypothesis.")
    print("Conclusion: There is a statistically significant association")
    print("between BMI category and diabetes outcome.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")
    print("Conclusion: There is insufficient evidence of an association")
    print("between BMI category and diabetes outcome.")
```
