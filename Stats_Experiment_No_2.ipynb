```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots in Colab
%matplotlib inline
sns.set_style("whitegrid")
```

```python
# Load the dataset
df = pd.read_csv("diabetes (1).csv")

# Display first 5 rows
df.head()
```

```python
# Dataset dimensions
print("Shape of Dataset:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# Dataset information
print("\nDataset Info:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Select numerical columns
numerical_cols = ['Pregnancies', 'Glucose', 'BloodPressure',
                  'SkinThickness', 'Insulin', 'BMI',
                  'DiabetesPedigreeFunction', 'Age']

# Mean
print("Mean")
print(df[numerical_cols].mean())

# Median
print("\nMedian")
print(df[numerical_cols].median())

# Mode
print("\nMode")
print(df[numerical_cols].mode().iloc[0])

# Minimum
print("\nMinimum")
print(df[numerical_cols].min())

# Maximum
print("\nMaximum")
print(df[numerical_cols].max())

# Variance
print("\nVariance")
print(df[numerical_cols].var())

# Standard Deviation
print("\nStandard Deviation")
print(df[numerical_cols].std())

# Overall descriptive statistics
print("\nDescriptive Statistics")
print(df[numerical_cols].describe())
```

```python
df[numerical_cols].hist(figsize=(15,10), bins=20)

plt.suptitle("Histograms of Numerical Variables", fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(15,8))

for i, col in enumerate(numerical_cols):
    plt.subplot(2,4,i+1)
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(x='Outcome', data=df)

plt.title("Frequency of Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.show()

print(df['Outcome'].value_counts())
```

```python
plt.figure(figsize=(7,5))

sns.scatterplot(x='Glucose', y='BMI', hue='Outcome', data=df)

plt.title("Glucose vs BMI")
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.scatterplot(x='Age', y='Glucose', hue='Outcome', data=df)

plt.title("Age vs Glucose")
plt.show()
```

```python
selected_columns = ['Glucose', 'BMI', 'Age', 'BloodPressure', 'Outcome']

sns.pairplot(df[selected_columns], hue='Outcome')

plt.show()
```
